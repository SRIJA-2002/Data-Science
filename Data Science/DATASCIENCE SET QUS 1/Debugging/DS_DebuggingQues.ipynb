```python
# 3
import numpy as np
from scipy.stats import zscore

data = [1, 2, 3, 100, 5, 6, 7, 8, 9]
#Z score for calculation
z_scores=zscore(data)
#abc for detect the outliers both directions
outliers = [data[i] for i in range(len(data)) if abs(z_scores[i]) > 2]
print(outliers)
```

```python
import numpy as np
from scipy.stats import zscore

data = [1, 2, 3, 100, 5, 6, 7, 8, 9]
median = np.median(data)

data = [median if abs(zscore(x)) > 2 else x for x in data]
```

```python
# 4
import numpy as np
from scipy.stats import zscore

data = [1, 2, 3, 100, 5, 6, 7, 8, 9]
median = np.median(data)
# calculate z score for full data
z_score=zscore(data)

data = [median if abs(z_score[i]) > 2 else x for i,x in enumerate (data)]
print(data)
```
