<div style="background-color:#FFFBEB; padding:12px; border-radius:8px; border-left:4px solid #F59E0B;">

<b>Question 1:</b><br>
How can you detect <b>multicollinearity</b> in this model?

</div>

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Real Estate Model Data (Example)
data = { 'House_Size': [1000, 1500, 2000, 2500, 3000, 1200, 1800],
    'Number_of_Rooms': [2, 3, 3, 4, 5, 2, 3],
    'Number_of_Bathrooms': [1, 2, 2, 3, 3, 1, 2]
}
df = pd.DataFrame(data)

# METHOD 1: Correlation Matrix (Easiest to detect)
print("Method 1: Correlation Matrix")
print(df.corr())
print("\nIf correlation > 0.8, then Multicollinearity exists!\n")

# METHOD 2: PAIR PLOT
plt.figure(figsize=(8,6))
sns.pairplot(df, diag_kind='kde', plot_kws={'color':'#21467E'})
plt.suptitle("Pairplot - Multicollinearity Check", y=1.02, fontsize=14, color='#21467E', fontweight='bold')
plt.show()

# METHOD 3: VIF Explanation in Output
print("Method 2: VIF - Variance Inflation Factor")
print("VIF > 5 means Multicollinearity detected!")
print("In this model, House Size, Rooms, Bathrooms are highly correlated with each other.")
```

<h1 style="color:#1A237E; font-size:28px; font-weight:bold; margin-bottom:5px;">Scenario 5: Medicine Effectiveness Study</h1>
<h3 style="color:#B71C1C; font-size:18px; margin-top:0px; font-weight:500;">for a Pharmaceutical Company</h3>

</center>

<div style="background-color:#F5F7FF; padding:20px; border-radius:12px; border-left:6px solid #1A237E; margin-top:20px;">

<b style="font-size:16px;">📦 Problem Statement:</b><br>
<span style="font-size:14px;">
A company made a new medicine to lower blood pressure. They gave it to one group and gave a fake pill (placebo) to another group.
</span>

<br><br>
<span style="font-size:14px;">
They want to check if the new medicine actually lowers blood pressure compared to placebo.
</span>

</div>

<br>

<div style="background-color:#FFF8E1; padding:12px 20px; border-radius:8px;">

<b style="color:#B71C1C;">📝 Question 1:</b> <span style="font-size:15px;">How can the company check if the new medicine works?</span>

</div>

```python
import numpy as np
from scipy import stats

# Sample Data 
# BP after taking medicine - Lower means effective
medicine_group = np.array([120, 118, 122, 115, 119, 121, 117, 116, 118, 120]) 
placebo_group = np.array([135, 138, 140, 132, 136, 134, 139, 137, 135, 133]) 

print(f"Average BP Medicine Group: {np.mean(medicine_group):.1f}")
print(f"Average BP Placebo Group: {np.mean(placebo_group):.1f}")

# STEP 1: Hypothesis
print("H0: Medicine has no effect (Both groups same)")
print("H1: Medicine is effective (Medicine group has lower BP)")

# STEP 2: Independent T-Test - Why? 2 Different Groups!
t_stat, p_value = stats.ttest_ind(medicine_group, placebo_group)

print(f"\nT-Value: {t_stat:.3f}")
print(f"P-Value: {p_value:.6f}")

# STEP 3: Decision
if p_value < 0.05:
    print("\n RESULT: P < 0.05 -> Medicine WORKS! Reject H0")
else:
    print("\n RESULT: P > 0.05 -> Medicine DOES NOT WORK")
```

<h1 style="color:#1A237E; font-size:28px; font-weight:bold; margin-bottom:5px;">Scenario 6: Identifying Outliers in Sales Data</h1>
<h3 style="color:#B71C1C; font-size:18px; margin-top:0px; font-weight:500;">for Sales Performance Analysis</h3>

</center>

<div style="background-color:#F5F7FF; padding:20px; border-radius:12px; border-left:6px solid #1A237E; margin-top:20px;">

<b style="font-size:16px;">📦 Problem Statement:</b><br>
<span style="font-size:14px;">
A company wants to find any unusual spikes in sales. They want to identify outliers in their sales data.
</span>

</div>

<br>

<div style="background-color:#FFF8E1; padding:12px 20px; border-radius:8px;">

<b style="color:#B71C1C;">📝 Question 1:</b> <span style="font-size:15px;">How can the company detect outliers in their sales data?</span>

</div>

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Sample Sales Data - in Lakhs (with some unusual spikes as outliers)
sales_data = np.array([10, 12, 11, 13, 12, 10, 11, 12, 50, 13, 11, 12, 10, 55, 12, 11, 13, 12])

df = pd.DataFrame(sales_data, columns=['Sales'])

print("Sales Data:")
print(df['Sales'].values)

# Method 1: IQR Method - BEST for Sales Data
Q1 = df['Sales'].quantile(0.25)
Q3 = df['Sales'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"\nQ1 (25%): {Q1}")
print(f"Q3 (75%): {Q3}")
print(f"IQR: {IQR}")
print(f"Lower Bound: {lower_bound:.2f}")
print(f"Upper Bound: {upper_bound:.2f}")

outliers = df[(df['Sales'] < lower_bound) | (df['Sales'] > upper_bound)]
print(f" Outliers Detected (Unusual Spikes): \n{outliers['Sales'].values}")
print(f"Total Outliers: {len(outliers)}")

# Visualization - Boxplot (Same Blue Color #21467E)
plt.figure(figsize=(8,4))

plt.subplot(1,2,1)
sns.boxplot(y=df['Sales'], color="#21467E")
plt.title("Boxplot - Outlier Detection", color="#1A237E", fontweight='bold')
plt.ylabel("Sales (in Lakhs)")

plt.subplot(1,2,2)
sns.histplot(df['Sales'], kde=True, color="#21467E")
plt.title("Sales Distribution", color="#1A237E", fontweight='bold')
plt.xlabel("Sales")

plt.tight_layout()
plt.show()
```

Detect outliers in sales data, we use the IQR method and Boxplot. I calculated Q1, Q3 and IQR. The normal sales range is 8.5 to 15.5 Lakhs. Values of 50 and 55 are above the upper bound, so they are detected as outliers or unusual spikes.

<h3 style="color:#B71C1C; font-size:18px; margin-top:0px; font-weight:500;">for Restaurant Survey Analysis</h3>

</center>

<div style="background-color:#F5F7FF; padding:20px; border-radius:12px; border-left:6px solid #1A237E; margin-top:20px;">

<b style="font-size:16px;">📦 Problem Statement:</b><br>
<span style="font-size:14px;">
A restaurant conducted a survey to rate customer satisfaction on a scale of 1 to 5:<br>
<b>[5, 4, 4, 5, 3, 4, 5, 2, 4, 3]</b>
</span>

</div>

<br>

<div style="background-color:#FFF8E1; padding:12px 20px; border-radius:8px;">

<b style="color:#B71C1C;">📝 Question 1:</b> <span style="font-size:15px;">How can the restaurant summarize the overall satisfaction?</span>

</div>

```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns

# Customer Satisfaction Ratings (1 to 5)
ratings = np.array([5, 4, 4, 5, 3, 4, 5, 2, 4, 3])

print(f"Ratings: {ratings}")

# BEST METHOD: Measures of Central Tendency (Mean, Median, Mode)
mean_val = np.mean(ratings)
median_val = np.median(ratings)
mode_val = stats.mode(ratings, keepdims=True)[0][0]

print(f"\n--- Overall Satisfaction Summary ---")
print(f"Mean (Average): {mean_val:.2f} / 5")
print(f"Median (Middle Value): {median_val} / 5")
print(f"Mode (Most Frequent): {mode_val} / 5")

# Frequency Count
unique, counts = np.unique(ratings, return_counts=True)
print(f"\nFrequency of Ratings:")
for u, c in zip(unique, counts):
    print(f"Rating {u}: {c} customers")

# Visualization - Countplot (Same Color #21467E)
plt.figure(figsize=(8,4))

plt.subplot(1,2,1)
sns.countplot(x=ratings, palette=["#21467E"]*len(unique))
plt.title("Customer Rating Frequency", color="#1A237E", fontweight='bold')
plt.xlabel("Satisfaction Rating (1-5)")
plt.ylabel("Number of Customers")

plt.subplot(1,2,2)
plt.pie(counts, labels=[f"Rating {u}" for u in unique], autopct='%1.1f%%', colors=sns.color_palette("Blues", len(unique)))
plt.title("Satisfaction Distribution", color="#1A237E", fontweight='bold')

plt.tight_layout()
plt.show()
```

To summarize overall satisfaction, we use Mean, Median and Mode. Mean is 3.9/5, Median is 4 and Mode is 4. Most customers gave rating 4, so overall satisfaction is good, average 3.9 out of 5.
