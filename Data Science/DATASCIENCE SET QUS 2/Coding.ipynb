```python
Question 1 :
An airline tracks flight delays (in minutes) for 20 flights. Analyze the flight delays to calculate percentiles, detect outliers, and evaluate the overall distribution.
DataSet:
delays = [15, 30, 45, 20, 25, 100, 5, 60, 35, 50, 
          120, 80, 10, 75, 90, 200, 55, 40, 70, 150]
Expected Output:

Percentiles (10th, 25th, 50th, 75th, 90th): [ 14.5   28.75  52.5   82.5  123.  ]
IQR: 53.75
Outliers: [200]
 
```

```python
import numpy as np
import matplotlib.pyplot as plt

# DataSet
delays = [15, 30, 45, 20, 25, 100, 5, 60, 35, 50,
          120, 80, 10, 75, 90, 200, 55, 40, 70, 150]

# 1. Calculate Percentiles (10th, 25th, 50th, 75th, 90th)
percentiles = np.percentile(delays, [10, 25, 50, 75, 90])
print(f"Percentiles (10th, 25th, 50th, 75th, 90th): {percentiles}")

# 2. Calculate IQR
Q1 = np.percentile(delays, 25)
Q3 = np.percentile(delays, 75)
IQR = Q3 - Q1
print(f"IQR: {IQR}")

# 3. Detect Outliers using IQR method
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers = [x for x in delays if x < lower_bound or x > upper_bound]
print(f"Outliers: {outliers}")

# 4. Box Plot for Flight Delays
plt.figure(figsize=(8, 5))
plt.boxplot(delays, vert=False, patch_artist=True, 
            boxprops=dict(facecolor='lightblue', color='blue'),
            medianprops=dict(color='red'))
plt.title('Box Plot for Flight Delays', fontsize=14, fontweight='bold')
plt.xlabel('Delay (in minutes)')
plt.grid(True, alpha=0.3)
plt.show()

```

```python
Question 2 :
A company wants to analyze the salary distribution of its employees to understand the central tendency and determine whether the data is skewed.
DataSet:
salaries = [30000, 32000, 35000, 37000, 40000, 42000, 45000, 47000, 50000, 55000,
            60000, 62000, 65000, 67000, 70000, 72000, 75000, 80000, 85000, 90000]
Expected Output:
Mean Salary: 56950.0
Median Salary: 57500.0
Mode Salary: 30000
The data is Left Skewed (Negative Skew)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde
#dataset
salaries = [30000, 32000, 35000, 37000, 40000, 42000, 45000, 47000, 50000, 55000,
            60000, 62000, 65000, 67000, 70000, 72000, 75000, 80000, 85000, 90000]


mean_salary = np.mean(salaries)
median_salary = np.median(salaries)
mode_salary = salaries[0]  

print(f"Mean Salary: {mean_salary}")
print(f"Median Salary: {median_salary}")
print(f"Mode Salary: {mode_salary}")

if mean_salary < median_salary:
    print("The data is Left Skewed (Negative Skew)")
else:
    print("The data is Right Skewed")

plt.figure(figsize=(8,5))


counts, bins, patches = plt.hist(salaries, bins=5, color='#8BBCD9', 
                                 edgecolor='black', linewidth=1.2)


kde = gaussian_kde(salaries)
x_range = np.linspace(30000, 90000, 200)
kde_scaled = kde(x_range) * len(salaries) * (bins[1]-bins[0])
plt.plot(x_range, kde_scaled, color='#1F5A8A', linewidth=2)

plt.title('Histogram of Employee Salaries', fontweight='bold')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
Question 3:
 A school wants to analyze the exam performance of students across three subjects: Mathematics, Science, and English. How can Data Science concepts be applied to understand their performance?
DataSet:
data = {
    'Student': ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J'],
    'Mathematics': [85, 78, 92, 88, 70, 95, 60, 80, 90, 76],
    'Science': [80, 85, 88, 70, 75, 92, 55, 82, 89, 78],
    'English': [78, 74, 85, 80, 68, 90, 50, 77, 83, 72]
}
Expected Output:
Descriptive Statistics
Histogram(graph)
Correlation Analysis(graph)
HeatMap(grapg)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = {
    'Student': ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J'],
    'Mathematics': [85, 78, 92, 88, 70, 95, 60, 80, 90, 76],
    'Science': [80, 85, 88, 70, 75, 92, 55, 82, 89, 78],
    'English': [78, 74, 85, 80, 68, 90, 50, 77, 83, 72]
}
df = pd.DataFrame(data)

# 1. Descriptive Statistics
print("Descriptive Statistics")
print(df[['Mathematics','Science','English']].describe())

# 2. Histogram
plt.figure()
plt.hist(df['Mathematics'], bins=5, alpha=0.6, label='Maths', edgecolor='black')
plt.hist(df['Science'], bins=5, alpha=0.6, label='Science', edgecolor='black')
plt.hist(df['English'], bins=5, alpha=0.6, label='English', edgecolor='black')
plt.title('Histogram(graph)')
plt.xlabel('Marks')
plt.legend()
plt.show()

# 3. Correlation Analysis
corr = df[['Mathematics','Science','English']].corr()
print("\nCorrelation Analysis")
print(corr)
plt.scatter(df['Mathematics'], df['Science'])
plt.title('Correlation Analysis(graph)')
plt.xlabel('Mathematics')
plt.ylabel('Science')
plt.show()

# 4. HeatMap
plt.imshow(corr, cmap='coolwarm')
plt.title('HeatMap(graph)')
plt.xticks([0,1,2], ['Maths','Science','English'])
plt.yticks([0,1,2], ['Maths','Science','English'])
plt.colorbar()
for i in range(3):
    for j in range(3):
        plt.text(j, i, f"{corr.values[i,j]:.2f}", ha='center', va='center')
plt.show()
```

```python
Question 4:
A pharmaceutical company conducted a clinical trial with two groups: one receiving medication and the other a placebo. How do you perform a hypothesis test to determine the effectiveness of the medication?
Dataset:
medication_group = [110, 115, 108, 102, 107, 99, 111, 104, 109, 101]
placebo_group = [120, 125, 130, 122, 128, 119, 124, 127, 123, 126]
Expected Output:
T-Statistic: -9.201427649220966
P-Value: 3.163912817600812e-08
Reject the null hypothesis: The medication is effective
```

```python
from scipy import stats

medication_group = [110, 115, 108, 102, 107, 99, 111, 104, 109, 101]
placebo_group = [120, 125, 130, 122, 128, 119, 124, 127, 123, 126]

# T-Test
t_stat, p_value = stats.ttest_ind(medication_group, placebo_group)

print(f"T-Statistic: {t_stat}")
print(f"P-Value: {p_value}")

# Hypothesis Testing 
alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis: The medication is effective")
else:
    print("Fail to reject the null hypothesis: No significant difference")
```

```python
Question 5 : A company conducted a customer satisfaction survey where customers rated their experience on a scale of 1 to 10. Analyze the survey results to calculate descriptive statistics and visualize the distribution of customer satisfaction ratings.
Sample DataSet:  
ratings = [8, 9, 7, 5, 6, 10, 9, 4, 7, 8, 
           6, 9, 10, 5, 8, 7, 6, 9, 10, 7]
Expected Output:
Mean Rating: 7.5
Median Rating: 7.5
Mode Rating: 7
Standard Deviation: 1.746424919657298
 
```

```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde

ratings = [8, 9, 7, 5, 6, 10, 9, 4, 7, 8,
           6, 9, 10, 5, 8, 7, 6, 9, 10, 7]

# 1. CALCULATIONS 
mean_rating = np.mean(ratings)
median_rating = np.median(ratings)
mode_rating = stats.mode(ratings, keepdims=True).mode[0]
std_dev = np.std(ratings)

print(f"Mean Rating: {mean_rating}")
print(f"Median Rating: {median_rating}")
print(f"Mode Rating: {mode_rating}")
print(f"Standard Deviation: {std_dev}")

# 2. GRAPH - Distribution 
plt.style.use('dark_background')
plt.figure(figsize=(8,5), facecolor='black')
plt.hist(ratings, bins=6, color='#C67A2E', edgecolor='white', linewidth=1.5)

# Orange KDE line
kde = gaussian_kde(ratings)
x_range = np.linspace(3.5, 10.5, 200)
plt.plot(x_range, kde(x_range)*len(ratings)*0.85, color='#FF9F1C', linewidth=2.5)

plt.title('Distribution of Customer Satisfaction', fontweight='bold', color='white')
plt.ylabel('Frequency', color='white')
plt.show()
```
